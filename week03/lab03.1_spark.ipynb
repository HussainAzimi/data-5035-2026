```python
from pyspark.sql.functions import *
df = (spark.read
      .option("header", "true")
      .option("inferSchema", "true")
      .csv("/Workspace/Users/hussainazimi.career@gmail.com/data-5035-2026/week03/hospitals.csv") 
      )

df.show()
```

```python
result = (
    df.groupBy("city")
      .agg(count("name").alias("num_hospitals"), sum("beds").alias("total_beds")
           )
)
result.show()
```

Scenario B: Healthcare Encounters

```
# Q1: Show each visit with patient and provider details.
# Join type: INNER JOIN.
# Assumptions: Every visit has valid PATIENT_ID and PROVIDER_ID, only visits with matching records appear.
import pandas as pd
from snowflake.snowpark.context import get_active_session
session = get_active_session()

visits = session.table('EXERCISE9.PUBLIC.VISITS').to_pandas()
patients = session.table('EXERCISE9.PUBLIC.PATIENTS').to_pandas()
providers = session.table('EXERCISE9.PUBLIC.PROVIDERS').to_pandas()

result = visits.merge(patients, on='PATIENT_ID').merge(providers, on='PROVIDER_ID')
result[['PATIENT_NAME', 'PROVIDER_NAME', 'VISIT_DATE']]
```

```
# Q2: Show all patients and any visits they may have had.
# Join type: LEFT JOIN (how='left').
# Assumptions: All patients appear, VISIT_ID is NaN for patients with no visits.
q2 = patients.merge(visits, on='PATIENT_ID', how='left')
q2[['PATIENT_NAME', 'VISIT_ID']]
```

```
# Q3: Show all providers and any visits they handled.
# Join type: LEFT JOIN (how='left').
# Assumptions: All providers appear; VISIT_ID is NaN for providers with no visits.
q3 = providers.merge(visits, on='PROVIDER_ID', how='left')
q3[['PROVIDER_NAME', 'VISIT_ID']]
```

```
# Q4: Find patients who have never had a visit.
# Join type: LEFT JOIN + filter on NULL (Anti-join pattern).
# Assumptions: A NaN VISIT_ID after left merge means the patient has zero visits.
q4 = patients.merge(visits, on='PATIENT_ID', how='left')
q4[q4['VISIT_ID'].isna()][['PATIENT_NAME']]
```

```
# Q5: Show visits handled by cardiology providers.
# Join type: INNER JOIN (chained) + filter after join.
# Assumptions: SPECIALTY column contains 'Cardiology' for cardiology providers, only matched visits appear.
q5 = visits.merge(patients, on='PATIENT_ID').merge(providers, on='PROVIDER_ID')
q5[q5['SPECIALTY'] == 'Cardiology'][['PATIENT_NAME', 'PROVIDER_NAME', 'VISIT_DATE']]
```

Scenario D: Marketing Campaign Performance

```
# Setup: Load all tables as Pandas DataFrames
import pandas as pd
from snowflake.snowpark.context import get_active_session
session = get_active_session()

customers = session.table('EXERCISE9.PUBLIC.CUSTOMERS_D').to_pandas()
sends = session.table('EXERCISE9.PUBLIC.CAMPAIGN_SENDS').to_pandas()
clicks = session.table('EXERCISE9.PUBLIC.CLICKS').to_pandas()
```

```
# Q1: Show all campaign sends with customer emails
# Join type: INNER JOIN (pd.merge default)
# Assumptions: Every send has a valid CUSTOMER_ID in CUSTOMERS_D; only matched sends appear
q1 = sends.merge(customers, on='CUSTOMER_ID')
q1[['EMAIL', 'CAMPAIGN_ID', 'SEND_DATE']]
```

```
# Q2: Identify whether each campaign send resulted in a click
# Join type: LEFT JOIN (how='left')
# Assumptions: All sends are included; a non-null CLICK_ID means the send was clicked (TRUE), otherwise FALSE
q2 = sends.merge(clicks, on='SEND_ID', how='left')
q2['CLICKED'] = q2['CLICK_ID'].notna()
q2[['SEND_ID', 'CLICKED']]
```

```
# Q3: Show all customers and any campaigns they received
# Join type: LEFT JOIN (how='left')
# Assumptions: All customers appear; CAMPAIGN_ID is NaN for customers who never received a campaign
q3 = customers.merge(sends, on='CUSTOMER_ID', how='left')
q3[['EMAIL', 'CAMPAIGN_ID']]
```

```
# Q4: Find campaign sends that were never clicked
# Join type: LEFT JOIN + filter on NULL (Anti-join pattern)
# Assumptions: A NaN CLICK_ID after left merge means the send was never clicked
q4 = sends.merge(clicks, on='SEND_ID', how='left')
q4[q4['CLICK_ID'].isna()][['SEND_ID']]
```

```
# Q5: Find customers who have never received a campaign
# Join type: LEFT JOIN + filter on NULL (Anti-join pattern)
# Assumptions: A NaN SEND_ID after left merge means the customer never received any campaign
q5 = customers.merge(sends, on='CUSTOMER_ID', how='left')
q5[q5['SEND_ID'].isna()][['EMAIL']]
```
